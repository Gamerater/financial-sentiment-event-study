## 4. Interpretation

- **Group t-test**: tests whether mean CAR differs between positive- and negative-sentiment events.
- **Correlation**: tests the strength/direction of the linear relationship between sentiment score and CAR across all events.
- **One-sample tests**: tests whether each sentiment group's mean CAR is different from zero.

A statistically significant result (p < 0.05) with the expected direction (positive sentiment leads to positive CAR, negative sentiment leads to negative CAR) supports the hypothesis that financial news sentiment correlates with short-term abnormal stock returns -- this is the core finding to report in the paper.
